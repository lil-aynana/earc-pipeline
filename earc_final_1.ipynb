## Standard RAG vs EARC — token reduction, EM, F1, latency

The cells below compare the **EARC method** (selection/compression) against a
**standard-RAG baseline** (full retrieved context stuffed into the LLM).
Run a single query first to confirm the plumbing, then the aggregated run.


```python
# --- Single-query smoke test: EARC vs standard-RAG baseline (no QA files needed) ---
from evaluation.evaluator import evaluate_one

qa = {
    "question_id": "smoke-1",
    "question": "When was the first iPhone released?",
    "answers": ["June 29, 2007", "2007"],
    "dataset": "smoke",
}

rec = evaluate_one(pipe, qa, run_baseline=True)

print("Question :", qa["question"])
print("Gold     :", qa["answers"])
print()
print(f"{'metric':<24}{'EARC':<28}{'Standard RAG':<28}")
print("-" * 80)
print(f"{'exact_match':<24}{rec.get('exact_match'):<28}{rec.get('baseline_exact_match'):<28}")
print(f"{'f1':<24}{round(rec.get('f1', 0), 4):<28}{round(rec.get('baseline_f1', 0), 4):<28}")
print(f"{'contains_gold':<24}{rec.get('contains_gold'):<28}{rec.get('baseline_contains_gold'):<28}")
print(f"{'context_tokens':<24}{rec.get('selected_tokens'):<28}{rec.get('baseline_context_tokens'):<28}")
print(f"{'latency_sec':<24}{round(rec.get('latency_sec', 0), 3):<28}{round(rec.get('baseline_latency_sec', 0), 3):<28}")
print("-" * 80)
print(f"token_reduction     : {rec.get('token_reduction')} tokens "
      f"({round(rec.get('token_reduction_pct', 0), 1)}% smaller context)")
print()
print("EARC answer     :", rec.get("answer"))
print("Baseline answer :", rec.get("baseline_answer"))
if "baseline_error" in rec:
    print("BASELINE ERROR  :", rec["baseline_error"])
```

```python
# --- Aggregated run: EARC vs standard-RAG baseline over a QA sample ---
# Set QA_DIR to the folder holding your qa_*.pkl shards (e.g. on Drive).
from pathlib import Path
import glob

# Auto-locate a folder containing qa_*.pkl; override manually if needed.
QA_DIR = None
for cand in ["data/qa_pairs", "/content/drive/MyDrive"]:
    hits = glob.glob(f"{cand}/**/qa_*.pkl", recursive=True)
    if hits:
        QA_DIR = str(Path(hits[0]).parent)
        break
# QA_DIR = "/content/drive/MyDrive/your/qa_pairs"  # <- uncomment & set if not found
print("QA_DIR:", QA_DIR)
assert QA_DIR, "No qa_*.pkl shards found — set QA_DIR manually above."

from evaluation.evaluator import run_evaluation

# Start small (10-20) since run_baseline=True doubles the LLM calls.
results = run_evaluation(pipe, Path(QA_DIR), sample_size=20, run_baseline=True)
s = results["summary"]["overall"]

print(f"\nEvaluated: {results['summary']['n_evaluated']}  |  errors: {results['summary']['n_errors']}\n")
print(f"{'metric':<26}{'EARC':<18}{'Standard RAG':<18}")
print("-" * 62)
print(f"{'exact_match':<26}{round(s['exact_match'],4):<18}{round(s['baseline_exact_match'],4):<18}")
print(f"{'f1':<26}{round(s['f1'],4):<18}{round(s['baseline_f1'],4):<18}")
print(f"{'contains_gold':<26}{round(s['contains_gold'],4):<18}{round(s['baseline_contains_gold'],4):<18}")
print(f"{'mean_context_tokens':<26}{round(s['mean_selected_tokens'],1):<18}{round(s['baseline_mean_context_tokens'],1):<18}")
print(f"{'mean_latency_sec':<26}{round(s['mean_latency_sec'],3):<18}{round(s['baseline_mean_latency_sec'],3):<18}")
print("-" * 62)
print(f"mean token reduction : {round(s['token_reduction_pct'],1)}% smaller context vs standard RAG")
print(f"grounded_rate        : {round(s['grounded_rate'],4)}   refusal_rate: {round(s['refusal_rate'],4)}")
```

## Testing with a different LLM (Mistral)

EARC is LLM-agnostic. The generation backend and model are now switchable, so
we can run the exact same pipeline with **Mistral** instead of Llama 3 and
compare. First pull the model, then build a pipeline that targets it.


```python
!ollama pull mistral
```

```python
# Same pipeline, Mistral as the generator. Indexes/corpus are reused; only the
# generation backend's target model changes.
pipe_mistral = EARCPipeline(backend="ollama", llm_model="mistral")

mistral_result = pipe_mistral.run("Who invented the telephone?")
print_report(
    mistral_result["query"],
    mistral_result["selected_sentences"],
    mistral_result["generation"],
)
print("\nModel used:", mistral_result["generation"]["backend"])
```
